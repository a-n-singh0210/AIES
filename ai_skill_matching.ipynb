```python
class User:
    def __init__(self, user_id, skills_offered, skills_wanted):
        self.user_id = user_id
        self.skills_offered = skills_offered
        self.skills_wanted = skills_wanted

# Example user data
users = [
    User(user_id=1, skills_offered=["Python", "Machine Learning"], skills_wanted=["Data Science"]),
    User(user_id=2, skills_offered=["Java", "Spring"], skills_wanted=["Python"]),
    User(user_id=3, skills_offered=["Data Science", "Statistics"], skills_wanted=["Machine Learning"]),
]
```

```python
import nltk
from nltk.corpus import wordnet


nltk.download('wordnet')

def normalize_skills(skills):
    normalized_skills = set()
    for skill in skills:
        #  synonyms using WordNet
        synonyms = wordnet.synsets(skill)
        normalized_skills.add(skill.lower())  # original skill
        for syn in synonyms:
            normalized_skills.add(syn.lemmas()[0].name().lower())  #first synonym
    return normalized_skills

# Normalize skills for each user
for user in users:
    user.skills_offered = normalize_skills(user.skills_offered)
    user.skills_wanted = normalize_skills(user.skills_wanted)
```

```python
def calculate_compatibility(user1, user2):
    offered_intersection = user1.skills_offered.intersection(user2.skills_wanted)
    wanted_intersection = user1.skills_wanted.intersection(user2.skills_offered)

    #Jaccard similarity
    union = user1.skills_offered.union(user2.skills_wanted)
    if not union:
        return 0
    score = (len(offered_intersection) + len(wanted_intersection)) / len(union)
    return score
```

```python
def match_users(users):
    matches = {}
    for user in users:
        matches[user.user_id] = []
        for other_user in users:
            if user.user_id != other_user.user_id:  # Avoid matching with self
                score = calculate_compatibility(user, other_user)
                matches[user.user_id].append((other_user.user_id, score))

    # Sort matches by compatibility score
    for user_id in matches:
        matches[user_id] = sorted(matches[user_id], key=lambda x: x[1], reverse=True)

    return matches

# Get matches
user_matches = match_users(users)

# Display matches
for user_id, match_list in user_matches.items():
    print(f"User  {user_id} matches:")
    for match in match_list:
        print(f"  User {match[0]} with score: {match[1]:.2f}")
```
